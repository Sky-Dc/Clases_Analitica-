```python
df = df.drop_duplicates()

print("Duplicados eliminados")
```

```python
columnas_numericas = df.select_dtypes(include=np.number).columns

columnas_numericas
```

```python
for columna in columnas_numericas:
    df[columna] = df[columna].fillna(df[columna].mean())

print("Valores numéricos corregidos")
```

```python
columnas_categoricas = df.select_dtypes(include='object').columns

columnas_categoricas
```

```python
for columna in columnas_categoricas:
    moda = df[columna].mode()[0]
    df[columna] = df[columna].fillna(moda)

print("Valores categóricos corregidos")
```

Transformacion de datos

```python
df["Grupo_Edad"] = pd.cut(
    df["Age"],
    bins=[0,18,25,35,45,60,100],
    labels=[
        "Menor",
        "Joven",
        "Adulto Joven",
        "Adulto",
        "Maduro",
        "Adulto Mayor"
    ]
)

df[["Age","Grupo_Edad"]].head()
```

CLASIFICACIÓN DEL CGPA

```python
df["Clasificacion_CGPA"] = pd.cut(
    df["CGPA"],
    bins=[0,5,7,8.5,10],
    labels=[
        "Bajo",
        "Regular",
        "Bueno",
        "Excelente"
    ]
)

df[["CGPA","Clasificacion_CGPA"]].head()
```

 TRANSFORMACIÓN DE HORAS DE SUEÑO

```python
df["Horas_Sueño"] = df["Sleep Duration"].map({
    "Less than 5 hours": 4,
    "5-6 hours": 5.5,
    "7-8 hours": 7.5,
    "More than 8 hours": 9
})

# Convertir a valores numéricos

df["Horas_Sueño"] = pd.to_numeric(
    df["Horas_Sueño"],
    errors="coerce"
)

df[["Sleep Duration","Horas_Sueño"]].head()
```

# OUTLIERS EN CGPA

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

# DETECTAR OUTLIERS

```python
outliers = df[
    (df["CGPA"] < limite_inferior) |
    (df["CGPA"] > limite_superior)
]

outliers
```

CANTIDAD DE OUTLIERS

```python
print("Outliers encontrados:", len(outliers))
```
