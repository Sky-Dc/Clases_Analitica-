```python
df = df.drop_duplicates()

print("Duplicados eliminados")
```

Seleccionamos las columnas numericas y rellenamos sus nulos con la media

```python
columnas_numericas = df.select_dtypes(include=np.number).columns

columnas_numericas
```

```python
for columna in columnas_numericas:
    df[columna] = df[columna].fillna(df[columna].mean())

print("Valores numéricos corregidos")
```

Hacemos lo mismo, pero rellenando los vacios con la moda

```python
columnas_categoricas = df.select_dtypes(include='object').columns

columnas_categoricas
```

```python
for columna in columnas_categoricas:
    moda = df[columna].mode()[0]
    df[columna] = df[columna].fillna(moda)

print("Valores categóricos corregidos")
```

Transformacion de datos

# Creamos nuevas variables para enriquecer el analisis

Agrupamos la edad en rangos

```python
df["Grupo_Edad"] = pd.cut(
    df["Age"],
    bins=[0,18,25,35,45,60,100],
    labels=[
        "Menor",
        "Joven",
        "Adulto Joven",
        "Adulto",
        "Maduro",
        "Adulto Mayor"
    ]
)

df[["Age","Grupo_Edad"]].head()
```

# Clasificacion del CGPA

```python
df["Clasificacion_CGPA"] = pd.cut(
    df["CGPA"],
    bins=[0,5,7,8.5,10],
    labels=[
        "Bajo",
        "Regular",
        "Bueno",
        "Excelente"
    ]
)

df[["CGPA","Clasificacion_CGPA"]].head()
```

# Transformacion de horas de sueño

Transformamos los textos de horas de sueño a valores numericos reales

```python
df["Horas_Sueño"] = df["Sleep Duration"].map({
    "Less than 5 hours": 4,
    "5-6 hours": 5.5,
    "7-8 hours": 7.5,
    "More than 8 hours": 9
})


df["Horas_Sueño"] = pd.to_numeric(
    df["Horas_Sueño"],
    errors="coerce"
)

df[["Sleep Duration","Horas_Sueño"]].head()
```

# OUTLIERS EN CGPA

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("Límite inferior:", limite_inferior)
print("Límite superior:", limite_superior)
```

# DETECTAR OUTLIERS

```python
outliers = df[
    (df["CGPA"] < limite_inferior) |
    (df["CGPA"] > limite_superior)
]

outliers
```

Cantidad de Outliers

```python
print("Outliers encontrados:", len(outliers))
```

Distribución Proporcional de los Hábitos Alimenticios según la Depresión

```python
cross_tab = pd.crosstab(
    df['Depression'], df['Dietary Habits'], normalize='index'
) * 100

# Graficar las barras 
cross_tab.plot(
    kind='bar', stacked=True, figsize=(9, 5), colormap='Spectral', edgecolor='black'
)
plt.title(
    'Distribución Proporcional de los Hábitos Alimenticios según la Depresión'
)
plt.xlabel('Depresión (0 = No, 1 = Sí)')
plt.ylabel('Porcentaje (%)')
plt.xticks(rotation=0)
plt.legend(title='Hábitos Alimenticios', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

Ranking de Ciudades con Mayor Concentración de Casos de Depresión

```python
df_depresion = df[df['Depression'] == 1]
city_counts = df_depresion['City'].value_counts().head(10).reset_index()
city_counts.columns = ['City', 'Casos']

plt.figure(figsize=(10, 6))
sns.barplot(
    data=city_counts,
    x='Casos',
    y='City',
    palette='Blues_r',
    hue='City',
    legend=False,
)
plt.title('Ranking de Ciudades con Mayor Número de Casos de Depresión')
plt.xlabel('Número de Casos')
plt.ylabel('Ciudad')
plt.tight_layout()
plt.show()
```

Nivel de promedio de estres financiero segun el Grado Academico 

```python
df_stress = (
    df.groupby('Degree')['Financial Stress'].mean().reset_index()
)  # O 'Profession' según tu columna
df_stress = df_stress.sort_values(by='Financial Stress', ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=df_stress,
    x='Degree',
    y='Financial Stress',
    palette='Reds_r',
    hue='Degree',
    legend=False,
)
plt.title('Nivel Promedio de Estrés Financiero por Grado Académico')
plt.xlabel('Grado Académico / Carrera')
plt.ylabel('Estrés Financiero Promedio')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
